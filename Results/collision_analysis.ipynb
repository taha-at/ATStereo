```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("trajectorySeparation.csv")
```

```python
data.describe()
```

```python
d_entry = data["Distance_Entry"]
d_target = data["Distance_Target"]
```

```python
std = d_entry.std()
average = d_entry.mean()

print(f"Entry distances {average} ± {std}")
```

```python
std = d_target.std()
average = d_target.mean()

print(f"Target distances {average} ± {std}")
```

```python
subjects = data["Subject"]
for i, sub in enumerate(subjects):
    print(f"Sub {sub} entry distance: {d_entry[i]} and target distance: {d_target[i]} and all are above 5 mm")
```
